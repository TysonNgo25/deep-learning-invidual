### Bài 7 Các hàm thống kê trên Tensor


```python
a = torch.randn(3, 4)
print("Tensor a:\n", a)
print("\nMean theo cột dim=0:\n", torch.mean(a, dim=0))
print("\nStandard deviation theo cột dim=0:\n", torch.std(a, dim=0))
print("\nCumulative sum theo cột dim=0:\n", torch.cumsum(a, dim=0))
print("\nMean theo hàng dim=1:\n", torch.mean(a, dim=1))
print("\nStandard deviation theo hàng dim=1:\n", torch.std(a, dim=1))
print("\nCumulative sum theo hàng dim=1:\n", torch.cumsum(a, dim=1))
```

### Bài 8 Autograd cơ bản


```python
x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()
print("Với y = x^2")
print("x =", x.item())
print("y =", y.item())
print("x.grad =", x.grad.item())
print("Đạo hàm thủ công 2x =", 2 * x.item())

x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2**2 + 2 * x2 + 1
y2.backward()
print("\nVới y = 3x^2 + 2x + 1")
print("y =", y2.item())
print("x.grad =", x2.grad.item())
print("Đạo hàm thủ công 6x + 2 =", 6 * x2.item() + 2)
```

**Nhận xét / trả lời:** Với y = x^2, dy/dx = 2x; tại x = 3.6 thì bằng 7.2, trùng với x.grad. Với y = 3x^2 + 2x + 1, dy/dx = 6x + 2; tại x = 3.6 bằng 23.6, trùng với gradient do autograd tính.


### Bài 9 Gradient Descent tìm hệ số đa thức


```python
import numpy as np
import torch
np.random.seed(3123411205)
polynomial = np.poly1d([1, 2, 3])
n = 40
X = np.random.randn(n, 1) * 5
Y = polynomial(X)
XX = np.hstack([X * X, X, np.ones_like(X)])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.03)
print("Initial coefficients:", w.detach().flatten().numpy())
for epoch in range(2500):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean((y - y_pred) ** 2)
    mse.backward()
    optimizer.step()
print("Learned coefficients:", w.detach().flatten().numpy())
print("Target coefficients:", np.array([1., 2., 3.]))
print("Final MSE:", mse.item())
```

**Nhận xét / trả lời:** Các hệ số học được xấp xỉ [1, 2, 3], là hệ số đã dùng để sinh dữ liệu. optimizer.zero_grad() xóa gradient tích lũy từ vòng lặp trước; mse.backward() lan truyền ngược để tính gradient của MSE theo w; optimizer.step() cập nhật w theo thuật toán NAdam để giảm MSE.

**Nhận xét / trả lời:** Các giá trị thay vào bốn phương trình cho kết quả xấp xỉ 9, 1, 8 và 2. Hệ có vô số nghiệm vì một phương trình phụ thuộc các phương trình còn lại: từ A+B=9, A+C=8 và B-D=2 suy ra C-D=1. Có thể đặt A=t, khi đó B=9-t, C=8-t và D=7-t.
